```python
# import python libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# import csv file
df=pd.read_csv('/content/Diwali Sales Data.csv', encoding='latin1')
df
```

```python
#drop unrelated/blank columns

df.drop(['Status', 'unnamed1'], axis=1, inplace=True)
```

Exploratory Data Analysis

1. UNIVARIATE ANALYSIS (Single Variable Distributions)

```python
# Plot 1.1: Age Distribution Curve (Continuous Variable Tracker)
fig, ax = plt.subplots(figsize=(9, 5))
sns.histplot(data=df,x="Age",kde=True,bins=15,ax=ax)
ax.set_title("1.1 Customer Baseline Age Profile Distribution", pad=15)
ax.set_xlabel("Age Profile (Years)")
ax.set_ylabel("Transaction Count")
plt.show()
```

```python
# Plot 1.2: Sector Employment Volume Ranking (Categorical Frequency Distribution)
fig, ax = plt.subplots(figsize=(10, 5))
order_rank = df["Occupation"].value_counts().index
sns.countplot(data=df, y="Occupation", order=order_rank, palette="flare", ax=ax)
ax.set_title("1.2 Order Inflow Volume Rank by Customer Occupation Group", pad=15)
ax.set_xlabel("Cumulative Successful Order Inflow Volume")
ax.set_ylabel("Industrial Occupation Sector")
plt.show()
```

2. BIVARIATE ANALYSIS (Dual Variable Correlation & Spread)

```python
# Plot 2.1: Age Group Segment Spread Box Profile (Continuous vs. Categorical Analysis)
fig, ax = plt.subplots(figsize=(10, 5))
age_bracket_order = ["0-17", "18-25", "26-35", "36-45", "46-50", "51-55", "55+"]
sns.boxplot(data=df,x="Age Group",y="Amount",ax=ax)
ax.set_title("2.1 Transaction Value Threshold Spread Across Age Group Cohorts", pad=15)
ax.set_xlabel("Demographic Age Classification Bracket")
ax.set_ylabel("Individual Gross Order Amount (INR)")
plt.show()
```

```python
# Plot 2.2: State-wise Aggregated Capital Flow Volume (Categorical Aggregation Profile)
fig, ax = plt.subplots(figsize=(11, 5))
state_revenue = (
    df.groupby("State")["Amount"].sum().reset_index().sort_values("Amount", ascending=False)
)
sns.barplot(data=state_revenue, x="Amount", y="State", palette="magma", ax=ax)
ax.set_title("2.2 Gross Retail Revenue Contribution by Regional State Source", pad=15)
ax.set_xlabel("Total Consolidated Revenue Yield")
ax.set_ylabel("Regional State Domain")
sns.despine(trim=True)
plt.tight_layout()
plt.show()
```

3. MULTIVARIATE ANALYSIS (Complex Pattern & Matrix Interactions)

```python
# Plot 3.1: Generographic Expenditure Density Breakdown via Dual Categorical Conditioning
fig, ax = plt.subplots(figsize=(11, 5))
sns.barplot(data=df,x="Age Group",y="Amount",hue="Gender",ax=ax)
ax.set_title(
    "3.1 Revenue Distribution Mapping Across Intersecting Age and Gender Profiles",
    pad=15,
)
ax.set_xlabel("Age Classification Bracket")
ax.set_ylabel("Aggregated Revenue Generation")
ax.legend(title="Gender Marker", loc="upper right")
plt.show()
```

```python
# Plot 3.2: Metric Matrix Statistical Heatmap Index (Correlation Architecture Mapping)
matrix_blueprint = df.copy()
# Encode discrete non-numeric metric markers safely into numerical fields for alignment maps
matrix_blueprint["Gender_Numerical"] = matrix_blueprint["Gender"].map({"M": 0, "F": 1})
structural_metrics = ["Age", "Marital_Status", "Orders", "Amount", "Gender_Numerical"]
statistical_correlation = matrix_blueprint[structural_metrics].corr()

fig, ax = plt.subplots(figsize=(8, 6))
sns.heatmap(statistical_correlation,annot=True,cmap="mako",linewidths=0.75,square=True,ax=ax)
ax.set_title("3.2 Correlation Index Intercept Heatmap Matrix", pad=20)
plt.show()
```
